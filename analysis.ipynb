UC 1 : Data Ingestion

```python
import pandas as pd

# Load datasets
deliveries_df = pd.read_csv("data/deliveries.csv")
matches_df = pd.read_csv("data/matches.csv")

# Show first rows
print("Deliveries Data:")
display(deliveries_df.head())

print("Matches Data:")
display(matches_df.head())
```

```python
# Shape
print("Deliveries Shape:", deliveries_df.shape)
print("Matches Shape:", matches_df.shape)

# Columns
print("\nDeliveries Columns:\n", deliveries_df.columns)
print("\nMatches Columns:\n", matches_df.columns)

# Data Types
print("\nDeliveries Info:")
print(deliveries_df.info())

print("\nMatches Info:")
print(matches_df.info())
```

UC 2 : Data Cleaning and Validation

Step 1: Check Missing Values


```python
# Check missing values
print("Missing values in deliveries:\n")
print(deliveries_df.isnull().sum())

print("\nMissing values in matches:\n")
print(matches_df.isnull().sum())
```

Step 2: Handle Missing Values

```python
# Matches dataset cleaning
matches_df['winner'] = matches_df['winner'].fillna("No Result")
matches_df['player_of_match'] = matches_df['player_of_match'].fillna("Unknown")
matches_df['city'] = matches_df['city'].fillna("Unknown")

# Deliveries dataset cleaning (UPDATED column names)
deliveries_df = deliveries_df.dropna(subset=['match_id', 'batter', 'bowler'])
```

Step 3: Fix Data type

```python
# Convert IDs to int
deliveries_df['match_id'] = deliveries_df['match_id'].astype(int)
matches_df['id'] = matches_df['id'].astype(int)

# Ensure numeric columns
deliveries_df['total_runs'] = pd.to_numeric(deliveries_df['total_runs'])
deliveries_df['batsman_runs'] = pd.to_numeric(deliveries_df['batsman_runs'])
```

Step 4: Validate Match IDs

```python
invalid_ids = set(deliveries_df['match_id']) - set(matches_df['id'])

print("Invalid Match IDs:", invalid_ids)
```

Step 5: Remove Invalid Data

```python
deliveries_df = deliveries_df[
    deliveries_df['match_id'].isin(matches_df['id'])
]
```

```python
deliveries_df['player_dismissed'] = deliveries_df['player_dismissed'].fillna("None")
```

UC 3 : Data Transformation

STEP 1: Create Total Runs Column

```python
# Recalculate total runs per ball
deliveries_df['calculated_total_runs'] = (
    deliveries_df['batsman_runs'] + deliveries_df['extra_runs']
)

# Check if matches existing total_runs
mismatch = (deliveries_df['total_runs'] != deliveries_df['calculated_total_runs']).sum()

print("Mismatch in total runs:", mismatch)
```

STEP 2: Standardize Column Names

```python
# Convert all columns to lowercase
deliveries_df.columns = deliveries_df.columns.str.lower()
matches_df.columns = matches_df.columns.str.lower()
```

STEP 3: Merge Both Datasets

```python
# Merge deliveries with matches
merged_df = deliveries_df.merge(
    matches_df,
    left_on="match_id",
    right_on="id",
    how="inner"
)

print("Merged Data Shape:", merged_df.shape)
display(merged_df.head())
```

STEP 4: Clean After Merge

```python
# Drop duplicate columns if needed
merged_df = merged_df.drop(columns=['id'], errors='ignore')
```

 Case 9: Boundary Percentage

```python
# Use Case 9: Boundary Percentage

# Total runs
total_runs = merged_df.groupby('batter')['batsman_runs'].sum()

# Boundary runs (4s & 6s)
boundary_runs = merged_df[
    merged_df['batsman_runs'].isin([4, 6])
].groupby('batter')['batsman_runs'].sum()

# Combine
boundary_pct_df = pd.concat([total_runs, boundary_runs], axis=1).reset_index()
boundary_pct_df.columns = ['batter', 'total_runs', 'boundary_runs']

# Fill missing
boundary_pct_df['boundary_runs'] = boundary_pct_df['boundary_runs'].fillna(0)

# Calculate %
boundary_pct_df['boundary_percentage'] = (
    boundary_pct_df['boundary_runs'] / boundary_pct_df['total_runs']
) * 100

# Filter
boundary_pct_df = boundary_pct_df[boundary_pct_df['total_runs'] > 500]

# Sort & display
boundary_pct_df = boundary_pct_df.sort_values(
    by='boundary_percentage', ascending=False
)

display(boundary_pct_df.head(10))
```

Case 10: Dot Ball Analysis

```python
# Use Case 10: Dot Ball Analysis

# Filter dot balls (total_runs == 0)
dot_balls_df = merged_df[merged_df['total_runs'] == 0]

# Count dot balls per bowler
dot_ball_counts = dot_balls_df.groupby('bowler').size().reset_index(name='dot_balls')

# Optional: filter bowlers with sufficient balls (avoid small sample)
bowler_balls = merged_df[merged_df['extras_type'] != 'wides'] \
    .groupby('bowler').size().reset_index(name='balls_bowled')

dot_ball_analysis = dot_ball_counts.merge(bowler_balls, on='bowler')

# Filter (important)
dot_ball_analysis = dot_ball_analysis[dot_ball_analysis['balls_bowled'] > 200]

# Sort
dot_ball_analysis = dot_ball_analysis.sort_values(
    by='dot_balls', ascending=False
)

display(dot_ball_analysis.head(10))
```

Case 11: Runs per Over Analysis

```python
# Use Case 11: Runs per Over Analysis

# Total runs per over (across all matches)
runs_per_over = merged_df.groupby('over')['total_runs'].sum()

# Number of balls per over (to calculate average)
balls_per_over = merged_df.groupby('over')['ball'].count()

# Average runs per over
avg_runs_per_over = (runs_per_over / balls_per_over) * 6

# Convert to DataFrame
runs_over_df = avg_runs_per_over.reset_index()
runs_over_df.columns = ['over', 'avg_runs_per_over']

# Sort (optional: descending to see highest scoring overs)
runs_over_df = runs_over_df.sort_values(
    by='avg_runs_per_over', ascending=False
)

display(runs_over_df)
```

Case 12: Powerplay Performance

```python
# Use Case 12: Powerplay Performance

# Filter powerplay overs (1–6)
powerplay_df = merged_df[merged_df['over'].between(1, 6)]

# Total runs per team in powerplay
team_powerplay_runs = powerplay_df.groupby('batting_team')['total_runs'].sum()

# Matches played by each team (for average calculation)
team_matches = merged_df.groupby('batting_team')['match_id'].nunique()

# Combine
powerplay_analysis = pd.concat([team_powerplay_runs, team_matches], axis=1).reset_index()
powerplay_analysis.columns = ['team', 'total_powerplay_runs', 'matches']

# Average runs per match in powerplay
powerplay_analysis['avg_powerplay_runs'] = (
    powerplay_analysis['total_powerplay_runs'] / powerplay_analysis['matches']
)

# Sort
powerplay_analysis = powerplay_analysis.sort_values(
    by='avg_powerplay_runs', ascending=False
)

display(powerplay_analysis.head(10))
```

Case 13: Death Overs Performance

```python
# Use Case 13: Death Overs Performance

# Filter death overs (16–20)
death_df = merged_df[merged_df['over'].between(16, 20)]

# 1. Team Performance

team_death_runs = death_df.groupby('batting_team')['total_runs'].sum()
team_matches = merged_df.groupby('batting_team')['match_id'].nunique()

death_team_analysis = pd.concat([team_death_runs, team_matches], axis=1).reset_index()
death_team_analysis.columns = ['team', 'total_death_runs', 'matches']

death_team_analysis['avg_death_runs'] = (
    death_team_analysis['total_death_runs'] / death_team_analysis['matches']
)

death_team_analysis = death_team_analysis.sort_values(
    by='avg_death_runs', ascending=False
)

# 2. Batter Performance

batter_death_runs = death_df.groupby('batter')['batsman_runs'].sum()
batter_balls = death_df[death_df['extras_type'] != 'wides'] \
    .groupby('batter').size()

death_batter_analysis = pd.concat([batter_death_runs, batter_balls], axis=1).reset_index()
death_batter_analysis.columns = ['batter', 'runs', 'balls']

# Filter meaningful players
death_batter_analysis = death_batter_analysis[death_batter_analysis['balls'] > 50]

# Strike rate in death overs
death_batter_analysis['strike_rate'] = (
    death_batter_analysis['runs'] / death_batter_analysis['balls']
) * 100

death_batter_analysis = death_batter_analysis.sort_values(
    by='strike_rate', ascending=False
)

# Display results
print("Top Teams in Death Overs:")
display(death_team_analysis.head(10))

print("\nTop Batters in Death Overs:")
display(death_batter_analysis.head(10))
```

Case 14: Run Distribution per Inning

```python
# Use Case 14: Run Distribution per Inning

# Total runs per inning (1st vs 2nd)
inning_runs = merged_df.groupby('inning')['total_runs'].sum().reset_index()

# Matches count (to calculate average)
matches_count = merged_df['match_id'].nunique()

# Average runs per inning
inning_runs['avg_runs_per_match'] = (
    inning_runs['total_runs'] / matches_count
)

# Rename inning values for clarity
inning_runs['inning'] = inning_runs['inning'].map({
    1: 'First Innings',
    2: 'Second Innings'
})

display(inning_runs)
```

Case 15: Toss Impact Analysis

```python
# Use Case 15: Toss Impact Analysis

# Total matches
total_matches = matches_df.shape[0]

# Matches where toss winner also won match
toss_and_match_win = matches_df[
    matches_df['toss_winner'] == matches_df['winner']
].shape[0]

# Toss win success percentage
toss_win_percentage = (toss_and_match_win / total_matches) * 100

# Create summary DataFrame
toss_analysis = pd.DataFrame({
    'total_matches': [total_matches],
    'toss_and_match_wins': [toss_and_match_win],
    'toss_win_percentage': [toss_win_percentage]
})

display(toss_analysis)
```

Case 16: Player of Match Contribution

```python
# Use Case 16: Player of Match Contribution

# Step 1: Runs per batter per match
batter_match_runs = merged_df.groupby(
    ['match_id', 'batter']
)['batsman_runs'].sum().reset_index()

# Step 2: Highest scorer in each match
top_scorers = batter_match_runs.loc[
    batter_match_runs.groupby('match_id')['batsman_runs'].idxmax()
]

top_scorers = top_scorers.rename(columns={'batter': 'top_scorer'})

# Step 3: Get player of match from matches_df
pom_df = matches_df[['id', 'player_of_match']].rename(
    columns={'id': 'match_id'}
)

# Step 4: Merge both
comparison_df = top_scorers.merge(pom_df, on='match_id')

# Step 5: Check if same
comparison_df['is_same'] = (
    comparison_df['top_scorer'] == comparison_df['player_of_match']
)

# Step 6: Calculate percentage
match_count = comparison_df.shape[0]
correct_awards = comparison_df['is_same'].sum()

accuracy = (correct_awards / match_count) * 100

# Final result
result_df = pd.DataFrame({
    'total_matches_checked': [match_count],
    'matches_where_pom_is_top_scorer': [correct_awards],
    'accuracy_percentage': [accuracy]
})

display(result_df)
```

Case 17: Venue-wise Analysis

```python
# Use Case 17: Venue-wise Analysis


# 1. Matches per venue
matches_per_venue = matches_df.groupby('venue')['id'].count().reset_index()
matches_per_venue.columns = ['venue', 'total_matches']

# 2. Total runs per venue

runs_per_venue = merged_df.groupby('venue')['total_runs'].sum().reset_index()


# 3. Merge both

venue_analysis = matches_per_venue.merge(runs_per_venue, on='venue')


# 4. Average runs per match

venue_analysis['avg_runs_per_match'] = (
    venue_analysis['total_runs'] / venue_analysis['total_matches']
)


# 5. Sort (high scoring venues first)

venue_analysis = venue_analysis.sort_values(
    by='avg_runs_per_match', ascending=False
)

display(venue_analysis.head(10))
```

Case 18: City-wise Scoring Trends

```python
# Use Case 18: City-wise Scoring Trends


# 1. Matches per city

matches_per_city = matches_df.groupby('city')['id'].count().reset_index()
matches_per_city.columns = ['city', 'total_matches']


# 2. Total runs per city

runs_per_city = merged_df.groupby('city')['total_runs'].sum().reset_index()


# 3. Merge both

city_analysis = matches_per_city.merge(runs_per_city, on='city')

# 4. Average runs per match

city_analysis['avg_runs_per_match'] = (
    city_analysis['total_runs'] / city_analysis['total_matches']
)

# 5. Handle unknown cities (optional)

city_analysis = city_analysis[city_analysis['city'] != "Unknown"]

# 6. Sort (high scoring cities first)

city_analysis = city_analysis.sort_values(
    by='avg_runs_per_match', ascending=False
)

display(city_analysis.head(10))
```

Case 19: Season-wise Run Trends

```python
# Use Case 19: Season-wise Run Trends


# 1. Get season info into merged_df

season_df = merged_df[['match_id', 'season', 'total_runs']]

# 2. Total runs per season

runs_per_season = season_df.groupby('season')['total_runs'].sum().reset_index()

# 3. Matches per season

matches_per_season = matches_df.groupby('season')['id'].count().reset_index()
matches_per_season.columns = ['season', 'total_matches']

# 4. Merge both

season_analysis = runs_per_season.merge(matches_per_season, on='season')

# 5. Average runs per match

season_analysis['avg_runs_per_match'] = (
    season_analysis['total_runs'] / season_analysis['total_matches']
)

# 6. Sort by season (chronological)

season_analysis = season_analysis.sort_values(by='season')

display(season_analysis)
```

Case 20: Winning Team Analysis

```python
# Use Case 20: Winning Team Analysis
# 1. Runs per team per match

team_scores = merged_df.groupby(
    ['match_id', 'batting_team']
)['total_runs'].sum().reset_index()

# 2. Get winning team based on runs

predicted_winner = team_scores.loc[
    team_scores.groupby('match_id')['total_runs'].idxmax()
]

predicted_winner = predicted_winner.rename(columns={
    'batting_team': 'predicted_winner',
    'total_runs': 'winning_runs'
})

# 3. Actual winner from matches_df

actual_winner = matches_df[['id', 'winner']].rename(
    columns={'id': 'match_id', 'winner': 'actual_winner'}
)

# 4. Merge both

winner_comparison = predicted_winner.merge(
    actual_winner, on='match_id'
)

# 5. Compare

winner_comparison['is_correct'] = (
    winner_comparison['predicted_winner'] == winner_comparison['actual_winner']
)

# 6. Accuracy

total_matches = winner_comparison.shape[0]
correct_predictions = winner_comparison['is_correct'].sum()

accuracy = (correct_predictions / total_matches) * 100

# 7. Result

result_df = pd.DataFrame({
    'total_matches': [total_matches],
    'correct_predictions': [correct_predictions],
    'accuracy_percentage': [accuracy]
})

display(result_df)
```
